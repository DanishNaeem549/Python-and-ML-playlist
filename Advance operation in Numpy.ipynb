```python
import numpy as np
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
```

```python
print("Sales Data:\n", sales_data)
```

```python
print(sales_data.shape)  # Output: (5, 5)
```

```python
print(sales_data[0:3,0:])
```

```python
print(np.sum(sales_data[:,1:],axis=0))
print(np.sum(sales_data[:,1:],axis=1))
```

```python
#minimum sales for each restaurant
min_sales = np.min(sales_data[:,1:], axis=1)
print("Minimum sales for each restaurant:", min_sales)

#maximum sales for each restaurant
max_sales = np.max(sales_data[:,1:], axis=1)
print("Maximum sales for each restaurant:", max_sales)
```

```python
maximum_sales = np.max(sales_data[:,1:], axis=0)
print("Maximum sales for each year:", maximum_sales)
```

```python
avg_sales = np.mean(sales_data[:,1:], axis=0)
print("Average sales for each year:", avg_sales)
```

```python
avg_sales = np.mean(sales_data[:,1:], axis=1)
print("Average sales for each year:", avg_sales)
```

```python
cumsum = np.cumsum(sales_data[:,1:], axis=1)
print("Cumulative sales for each restaurant:\n", cumsum)
```
