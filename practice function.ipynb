```python
def greater(a,b,c):
    if a > b and a > c:
        return a
    elif b > a and b > c:
        return b
    else:
        return c


print(greater(10, 20, 30))
    
```

```python
def sum(n):
    if n == 0:
        return 0
    else:
        return n + sum(n-1)


print(sum(5))
```

```python
def pattern(n):
    for i in range(n, -1, -1):
        for j in range(i+1):
            print("*", end="")
        print()


pattern(5)        
```

```python
def length_converter(length):

    return (f"{length * 2.54} cm")

print(length_converter(10))
```

```python
def remove_word(*agr):

    word = input("Enter a word to remove: ")
    if word in agr:
        agr.remove(word)
    print(agr)

list1=[1,2,3,4,5,6,"danish","Bilal"]
remove_word(list1)

```
