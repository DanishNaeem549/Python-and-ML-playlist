```python
# Question 1 

fruits = []
for i in range(7):
    fruit = input("Enter the name of a fruit: ")
    fruits.append(fruit)


print("The fruits you entered are:", fruits)


```

```python
# Question 2

marks = []
for i in range(6):
    mark = int(input("Enter the marks of student {}: ".format(i+1)))
    marks.append(mark)

marks.sort()
print("The marks you entered are:", marks)
```

```python
# Question 3

list1 = [1, 2, 3, 4, 5]
sum = 0

for i in range(len(list1)):
    sum += list1[i]

print("The sum of the elements in the list is:", sum)



```
