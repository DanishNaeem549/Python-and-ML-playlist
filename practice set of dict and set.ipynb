```python
# Question 5

dict={}

for i in range(5):
    name = input("Enter the name of student {}: ".format(i+1))
    marks = int(input("Enter the marks of student {}: ".format(i+1)))
    dict[name] = marks

print("The dictionary of students and their marks is:", dict)    
```
