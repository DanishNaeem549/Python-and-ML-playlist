```python
import pandas as pd
```

```python
# series is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). The axis labels are collectively referred to as the index. A pandas Series is like a fixed-size dict in that you can get and set values by index label.

# ceating the series from list
data = [1, 2, 3, 4, 5]
series_from_list = pd.Series(data)
print("Series from list:\n", series_from_list)
```

```python
# ceating the series from dictionary
data = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
series_from_dict = pd.Series(data)
print("Series from dictionary:\n", series_from_dict)
```

```python
# give index to the series
data = [10, 20, 30, 40, 50]
index = ['a', 'b', 'c', 'd', 'e']
series_with_index = pd.Series(data, index=index)
print("Series with custom index:\n", series_with_index)
```

```python
import numpy as np
```

```python
arr = np.array(series_with_index)
print("Numpy array from pandas series:\n", arr)
```

```python
# DataFrame is a two-dimensional labeled data structure with columns of potentially different types. You can think of it like a spreadsheet or SQL table, or a dict of Series objects. It is generally the most commonly used pandas object.

#creting the DataFrame from dictionary of list

data = {'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
        'age': [25, 30, 35, 40, 45],
        'city': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
        }

df=pd.DataFrame(data)
print("DataFrame from dictionary of lists:\n", df)        
```

```python
data = {'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
        'age': [25, 30, 35, 40, 45],
        'city': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
        }
index=['row1', 'row2', 'row3', 'row4', 'row5']

df=pd.DataFrame(data, index=index)
print("DataFrame from dictionary of lists:\n", df)
```

```python
#creating the DataFrame from list of dictionaries

data = [{'name': 'Alice', 'age': 25, 'city': 'New York'},
        {'name': 'Bob', 'age': 30, 'city': 'Los Angeles'},
        {'name': 'Charlie', 'age': 35, 'city': 'Chicago'},
        {'name': 'David', 'age': 40, 'city': 'Houston'},
        {'name': 'Eva', 'age': 45, 'city': 'Phoenix'}]

df = pd.DataFrame(data)
print("DataFrame from list of dictionaries:\n", df)
```
