```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student = Student("Danish", 21)

print(student.name)
print(student.age)
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Student: {self.name}, Age: {self.age}"

student = Student("Danish", 21)

print(student)
```

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value


num1 = Number(10)
num2 = Number(20)

print(num1 + num2)
```

```python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)


team = Team(["Ali", "Danish", "Ahmed"])

print(len(team))
```

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __eq__(self, other):
        return self.name == other.name


s1 = Student("Danish")
s2 = Student("Danish")

print(s1 == s2)
```
