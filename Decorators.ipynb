```python
def my_decorator(func):

    def wrapper():
        print("Before function runs")
        func()
        print("After function runs")

    return wrapper


@my_decorator
def hello():
    print("Hello, Danish!")


hello()
```

```python
def my_decorator(func):

    def wrapper(name):
        print("Function started")
        func(name)
        print("Function ended")

    return wrapper


@my_decorator
def greet(name):
    print(f"Hello {name}")


greet(str(input("Enter your name: ")))
```

```python
def decorator(func):

    def wrapper(*args, **kwargs):
        print("Function is running...")
        print(func(*args, **kwargs))
        print("Function finished!")

    return wrapper


@decorator
def add(a, b):
    return a + b


add(4,5)
```

```python
def login_required(func):

    def wrapper(username):
        if username == "admin":
            func(username)
        else:
            print("Access denied!")

    return wrapper


@login_required
def dashboard(username):
    print(f"Welcome {username}")


dashboard("admin")
dashboard("Danish")
```
