```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tqdm import tqdm_notebook
```

```python
df = pd.read_csv('/content/sample_submission.csv')
```

```python
# Building model
base_model = tf.keras.applications.MobileNetV2(input_shape=(224,224,3),include_top=False,weights='imagenet')
```

```python
# Summary of the model
base_model.summary()
```

```python
#creating the header for the base model
base_model.trainable = False
```

```python
# create the layer for model
base_model.output
Avg_model_lay = tf.keras.layers.AveragePooling2D()(base_model.output)
predication_layer = tf.keras.layers.Dense(units=1, activation="sigmoid")(Avg_model_lay)
```

```python
# create the model
model = tf.keras.models.Model(inputs = base_model.output, outputs = predication_layer)
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
gen_train = ImageDataGenerator()
gen_valid = ImageDataGenerator()
train = gen_train.flow_from_directory('/content/train',target_size=(224,224),batch_size=128,class_mode='binary')
valid = gen_valid.flow_from_directory('/content/test',target_size=(224,224),batch_size=128,class_mode='binary')
```

```python
model.fit_generator(train, epochs=3, validation_data=valid)
```
