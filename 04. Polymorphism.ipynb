```python
class Animal:
    def speak(self):
        return "Sound of the animal"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"


# fuction that demonstrates polymorphism
def animal_speak(Animal):
    print(Animal.speak())


dog=Dog()
cat= Cat()
print(dog.speak()) 
animal_speak(cat)
```

```python

class shape:
    def area(self):
        return " the are of the shape "

class Rectangle(shape):
    def __init__(self, name, width, height):
        self.name= name
        self.width= width
        self.height= height


    def area(self):
        return self.width * self.height    


class circle(shape):
    def __init__(self, name, radius):
            self.name= name
            self.radius= radius


    def area(self):
         return 3.14* self.radius*self.radius


# fuction that demonstrates polymorphism
def area_shape(shape):
    print(f"{shape.name} area is: {shape.area()}")



circl = circle("circle", 6)
print(circl.area())

rect = Rectangle("Rectangle", 4,6)
area_shape(rect)
    
```
