```python
def numbers():
    for i in range(1, 6):
        yield i


for number in numbers():
    print(number)
```

```python
def read_data(data):
    for item in data:
        yield item


data = ["User1", "User2", "User3", "User4"]

for user in read_data(data):
    print(user)
```

```python
dat = read_data(data)
```

```python
next(dat)  # Output: User1
```

```python
def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()
```

```python
file_path = 'large.txt'  # Replace with the path to your large file
for line in read_large_file(file_path):
    print(line)
```

```python
line = read_large_file(file_path)
```

```python
next(line)  # Output: User1
```
