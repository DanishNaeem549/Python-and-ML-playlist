```python
# list in Python is a collection of items that are ordered and changeable. Lists are written with square brackets.

numbers = [1, 2, 3, 4, 5]
friends = ["Alice", "Bob", "Charlie", "David", "Eve"]

print(friends[2])

friends[2]="Danish"   # list are mutable so we can change the value of an item in a list by referring to its index number.
print(friends[2])

#slicing in list
print(friends[1:4])  # prints items from index 1 to 3

# methods in list
friends.append("Frank")  # adds an item to the end of the list
print(friends)
friends.insert(2, "Grace")  # adds an item at the specified index
print(friends)
friends.sort()  # sorts the list in ascending order
print(friends)
friends.remove("Bob")  # removes the specified item from the list
print(friends)
friends.pop(3)  # removes the item at the specified index
print(friends)

list1 = numbers + friends  # concatenates two lists
print(list1)
```

```python
# tuple in Python is a collection of items that are ordered and unchangeable. Tuples are written with round brackets.

numbers = (1, 2, 3, 4, 5)
friends = ("Alice", "Bob", "Charlie", "David", "Eve")
print(friends[2])
print(friends[4])

print(friends[1:4])  # prints items from index 1 to 3


friends.count("Alice")  # returns the number of times a specified value occurs in a tuple
friends.index("Charlie")  # returns the index of the first occurrence of a specified value in a tuple

tuple1 = numbers + friends  # concatenates two tuples
print(tuple1)

tuple2 = numbers * 3  # repeats the tuple a specified number of times
print(tuple2)


print("Bob" in tuple1)  # returns True if the specified value is present in the tuple, otherwise False
print(len(tuple1))

```

```python
# set in Python is a collection of items that are unordered and unindexed. Sets are written with curly brackets.
 
numbers = {1, 2, 3, 4, 5,5,5,5}
friends = {"Alice", "Bob", "Charlie", "David", "Eve"}

empty_set = set()  # creates an empty set
print(type(empty_set))

print(numbers)  # prints the set, duplicates are removed

friends.add("Frank")  # adds an item to the set
print(friends)

print("Bob" in friends)  # returns True if the specified value is present in the set, otherwise False
print(len(friends))  # returns the number of items in the set
friends.remove("Alice")  # removes the specified item from the set
print(friends)
friends.pop()  # removes a random item from the set
print(friends)

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print(set1.union(set2))  # returns a set that contains all items from both sets
print(set1.intersection(set2))  # returns a set that contains only the items that are present in both sets




# print(friends[2])   this will raise an error because sets are unordered and unindexed
```

```python
# dict in Python is a collection of items that are unordered, changeable and indexed. Dictionaries are written with curly brackets, and they have keys and values.

numbers = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
marks = {"Alice": 90, "Bob": 85, "Charlie": 95, "David": 80, "Eve": 75}

print(marks)
print(type(marks))
print(marks["Charlie"])  # prints the value associated with the key "Charlie"

marks["Frank"] = 70  # adds a new key-value pair to the dictionary
print(marks)
marks["Alice"] = 95  # changes the value associated with the key "Alice"
print(marks)
marks.pop("Bob")  # removes the key-value pair with the specified key
print(marks)
print(len(marks))  # returns the number of key-value pairs in the dictionary

print(marks.keys())  # returns a list of all the keys in the dictionary
print(marks.values())  # returns a list of all the values in the dictionary
print(marks.items())  # returns a list of all the key-value pairs in the dictionary
print(marks.get("Charlie"))  # returns the value associated with the specified key


```
