```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = sns.load_dataset('tips')
```

```python
data
```

```python
# scatter plot
sns.scatterplot(x='total_bill',y='tip',data=data)
```

```python
# bar plot
sns.barplot(x='day',y='total_bill',data=data)
```

```python
# box plot
sns.boxplot(x='smoker',y='total_bill',data=data)
```

```python
# violin plot
sns.violinplot(x='sex',y='total_bill',data=data)
```

```python
# histogram
sns.histplot(data['size'],bins=10, kde=True)
```

```python
# Kde plot
sns.kdeplot(data['total_bill'],shade=True)
```

```python
# pair plot
sns.pairplot(data)
```
