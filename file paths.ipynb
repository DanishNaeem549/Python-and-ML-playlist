```python
path = "example.txt"

if os.path.exists(path):
    print("file exists in the folder")
else:
    print("file not exist")    
```

```python
path = "example.txt"

if os.path.isfile(path):
    print("It's a file")
elif os.path.isdir(path):
    print("It's a folder")     
else:
    print("file not exist") 
```
