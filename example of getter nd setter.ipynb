```python
class Student:
    def __init__(self, name, age):
        self._name = name
        self._age = age

    # Getter
    @property
    def age(self):
        return self._age

    # Setter
    @age.setter
    def age(self, value):
        if value >= 0:
            self._age = value
        else:
            print("Age cannot be negative")


# Create object
student = Student("Danish", 21)

# Getter
print(student.age)  # 21

# Setter
student.age = 22

# Getter again
print(student.age)  # 22
```
