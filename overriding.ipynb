```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __add__(self, other):
        return self.marks + other.marks


student1 = Student(80)
student2 = Student(90)

result = student1 + student2

print(result)
```

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __sub__(self, other):
        return self.value - other.value


num1 = Number(50)
num2 = Number(20)

print(num1 - num2)
```

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __mul__(self, other):
        return self.value * other.value


num1 = Number(10)
num2 = Number(5)

print(num1 * num2)
```

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __eq__(self, other):
        return self.marks == other.marks


s1 = Student(90)
s2 = Student(90)

print(s1 == s2)
```
