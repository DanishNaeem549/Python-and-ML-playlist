```python
# Inheritance in python (single child inheritance)

class car:
    def __init__(self, brand, model, year, color, enginetype):
        self.brand = brand
        self.model = model
        self.year = year
        self.color = color
        self.enginetype = enginetype

    def start(self):
        print(f"{self.brand} {self.model} has started.")

    def stop(self):
        print(f"{self.brand} {self.model} has stopped.")

    def accelerate(self):
        print(f"{self.brand} {self.model} is accelerating.")

    def driving(self):
        print(f"The person is driving {self.enginetype} car")

    def display_info(self):
        print("Car Information")
        print("----------------")
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Year :", self.year)
        print("Color:", self.color)

```

```python
class Tasla(car):
    def __init__(self, brand, model, year, color, enginetype, is_selfdriving):
        super().__init__(brand, model, year, color, enginetype)
        self.is_selfdriving = is_selfdriving

    def selfdrive(self):
        print(f"{self.brand} support self driving : {self.is_selfdriving}")
```

```python
tasla = Tasla("Talsa","Y",2026, "white", "electric", True)
tasla.selfdrive()
tasla.start()
tasla.display_info()
```

```python
# multiple child inheritance 


class Car:
    def drive(self):
        print("Driving on the road.")

    def stop(self):
        print("Car stopped.")


class Airplane:
    def fly(self):
        print("Flying in the sky.")

    def land(self):
        print("Airplane landed.")


class FlyingCar(Car, Airplane):
    def switch_mode(self):
        print("Switching from driving mode to flying mode.")


# Create object
vehicle = FlyingCar()

vehicle.drive()         # Inherited from Car
vehicle.stop()          # Inherited from Car
vehicle.fly()           # Inherited from Airplane
vehicle.land()          # Inherited from Airplane
vehicle.switch_mode()   # Own method
```

```python
class Car:
    def __init__(self, brand):
        self.brand = brand
        print(f"Car Brand: {self.brand}")


class Airplane:
    def __init__(self, wings):
        self.wings = wings
        print(f"Number of Wings: {self.wings}")


class FlyingCar(Car, Airplane):
    def __init__(self, brand, wings, model):
        Car.__init__(self, brand)          # Call Car constructor
        Airplane.__init__(self, wings)     # Call Airplane constructor

        self.model = model
        print(f"Model: {self.model}")

    def display(self):
        print("\nFlying Car Details")
        print("-------------------")
        print("Brand :", self.brand)
        print("Model :", self.model)
        print("Wings :", self.wings)


# Create object
fc = FlyingCar("Tesla", 2, "Model X Air")
fc.display()
```
