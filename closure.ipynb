```python
def welcome():
    msg = "Welcome to Python Programming"
    def inner():
        print("This is inner function")
        print(msg)
    
    return inner()



welcome()


                   # Output: This is inner function
                   #         Welcome to Python Programming
```

```python
def welcome(func):
    msg = "Welcome to Python Programming"
    def inner():
        print("This is inner function")
        print(msg)
        print(func())
    
    return inner()



welcome(lambda: "This is a sample function")
```

```python
def welcome(func,n):
    msg = "Welcome to Python Programming"
    def inner():
        print("This is inner function")
        print(msg)
        print(func(n))
    
    return inner()

factorial = lambda n: 1 if n == 0 else n * factorial(n - 1)

welcome(factorial,5)  # Output: This is inner function
```
