```python
import pandas as pd
```

```python
import numpy as np
```

```python
df = pd.read_csv('practice.csv')
```

```python
df
```

```python
# grouping by a single column and applying an aggregation function
groupdata_mean= df.groupby('Department')['Net_Amount'].mean()
print(groupdata_mean)
```

```python
df['Net_Amount'] = df['Net_Amount'].astype('int64')
```

```python
df
```

```python
groupdata_sum = df.groupby(['Department','City'])['Unit_Price'].sum()
print(groupdata_sum)
```

```python
groupdata_sum = df.groupby(['City','Department','Product'])['Unit_Price'].sum()
with open('groupdata_sum.csv', 'w') as f:
    groupdata_sum.to_csv(f)
```

```python
group_agg = df.groupby(['Department','City'])['Unit_Price'].agg([np.sum, np.mean, np.std])
print(group_agg)
```

```python
with open('group_agg.csv', 'w') as f:
    f.write(group_agg.to_csv())
```

```python
# marging and joining dataframes

df1 = pd.DataFrame({'key': ['A', 'B', 'C', 'D'],
                    'value': [1, 2, 3, 4]})

df2 = pd.DataFrame({'key': ['B', 'D', 'E', 'F'],
                    'value': [5, 6, 7, 8]})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1, df2, on='key', how='inner')  # inner join
```

```python
pd.merge(df1, df2, on='key', how='outer')  # outer join
```

```python
df3 =pd.merge(df1,df2, on='key', how='left')  # left join
```

```python
df3
```

```python
df3['salary'] = [1000, 2000, 3000, 4000]
```

```python
df3.rename(columns={'value_x': 'value_df1', 'value_y': 'value_df2'}, inplace=True)
```

```python
df3.isnull().sum()
```

```python
df3['value_df2'] = df3['value_df2'].fillna(df3['value_df2'].sum())
```
