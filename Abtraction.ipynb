```python
from abc import ABC, abstractmethod

class Car(ABC):

    @abstractmethod
    def start(self):
        pass


class BMW(Car):

    def start(self):
        print("BMW starts with a button")


class Toyota(Car):

    def start(self):
        print("Toyota starts with a key")


car1 = BMW()
car1.start()

car2 = Toyota()
car2.start()
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Dog says Woof")


class Cat(Animal):

    def sound(self):
        print("Cat says Meow")


dog = Dog()
dog.sound()

cat = Cat()
cat.sound()
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):

    def area(self):
        radius = 5
        return 3.14 * radius * radius


class Rectangle(Shape):

    def area(self):
        length = 10
        width = 5
        return length * width


circle = Circle()
print(circle.area())

rectangle = Rectangle()
print(rectangle.area())
```

```python
from abc import ABC, abstractmethod

class BankAccount(ABC):

    @abstractmethod
    def withdraw(self, amount):
        pass


class SavingsAccount(BankAccount):

    def withdraw(self, amount):
        print(f"Withdrawing {amount} from savings account")


class CurrentAccount(BankAccount):

    def withdraw(self, amount):
        print(f"Withdrawing {amount} from current account")


account = SavingsAccount()
account.withdraw(5000)
```
