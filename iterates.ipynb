```python
# Iterators: that allow you to traverse through all the elements of a collection, regardless of its specific implementation. In Python, an iterator is an object that implements the iterator protocol, which consists of the methods __iter__() and __next__(). The __iter__() method returns the iterator object itself, and the __next__() method returns the next value from the collection. When there are no more items to return, __next__() raises a StopIteration exception.


my_list = [1, 2, 3, 4, 5]

iterator = iter(my_list)
```

```python
try:
    print(next(iterator))  # Output: 1

except StopIteration:
    print("No more items in the iterator.")
```

```python
class EndOfIteratorError(Exception):
    pass


class MyNumbers:
    def __init__(self, max_number):
        self.max_number = max_number
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.max_number:
            raise EndOfIteratorError("No more numbers available")

        number = self.current
        self.current += 1

        return number


numbers = MyNumbers(5)

iterator = iter(numbers)

try:
    while True:
        print(next(iterator))

except EndOfIteratorError as e:
    print("Error:", e)
```
