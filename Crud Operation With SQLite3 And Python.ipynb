```python
import sqlite3
```

```python
connection = sqlite3.connect('my_database.db')
connection
```

```python
cursor = connection.cursor()
```

```python
cursor.execute('''CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age int,
    salary int 
)''')
connection.commit()
```

```python
cursor.execute("INSERT INTO users (name, age, salary) VALUES ('John Doe', 30, 50000)")
cursor.execute("INSERT INTO users (name, age, salary) VALUES ('Jane Smith', 25, 60000)")
connection.commit()
```

```python
cursor.execute('select * from users')
rows = cursor.fetchall()

for row in rows:
    print(row)
```
